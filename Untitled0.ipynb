```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
t=np.linspace(0, 2*np.pi, 500)
x=16*np.sin(t)**3
y=13*np.cos(t)-5*np.cos(2*t)-2*np.cos(3*t)-np.cos(4*t)
fig, ax=plt.subplots(figsize=(5,5))
ax.set_xlim(-25,25)
ax.set_ylim(-25,25)
ax.set_aspect("equal")
ax.axis("off")
# plt.figure(figsize=(6,6))
hati=ax.fill(x,y, facecolor="crimson",edgecolor="darkred",linewidth=3)[0]
def update(frame):
  skala=1+0.2*np.sin(frame/2)
  hati.set_xy(np.column_stack([x*skala, y*skala]))
  return hati,
ani=FuncAnimation(fig, update, frames=63,interval=50, blit=False)
plt.close()
HTML(ani.to_jshtml())
# plt.axis("equal")
# plt.axis("off")
```
